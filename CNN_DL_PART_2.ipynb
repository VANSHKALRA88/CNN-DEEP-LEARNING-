```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import Perceptron
from sklearn.metrics import accuracy_score , classification_report , confusion_matrix
from tensorflow.keras.models import Sequential
from sklearn.preprocessing import LabelEncoder
from sklearn.preprocessing import StandardScaler
```

```python
import pandas as pd
import numpy as np


train_df = pd.read_csv("/content/mnist_train.csv")
test_df  = pd.read_csv("/content/mnist_test.csv")


X_train = train_df.drop("label", axis=1).values
y_train = train_df["label"].values

X_test  = test_df.drop("label", axis=1).values
y_test  = test_df["label"].values

print(f"Train: {X_train.shape}, Test: {X_test.shape}")
```

```python
X_train=train_df.drop('label',axis=1).values
y_train=train_df['label'].values
X_test=test_df.drop('label',axis=1).values
y_test=test_df['label'].values
```

```python
X_train=X_train.astype("float32")/255.0
X_test=X_test.astype("float32")/255.0
```

```python
X_train_img=X_train.reshape(-1,28,28)
X_test_img=X_test.reshape(-1,28,28)
```

```python
X_train_img
```

```python
y_train_cat=to_categorical(y_train,10)
y_test_cat=to_categorical(y_test,10)
```

```python
perceptron = Sequential([
    Flatten(input_shape=(28,28)),
    Dense(10, activation="softmax")
])
```

```python
from sklearn import metrics
perceptron.compile(optimizer="sgd" , loss="categorical_crossentropy", metrics=["accuracy"])
```

```python
history_percp = perceptron.fit(X_train_img, y_train_cat, epochs=5,
batch_size=32, validation_data=(X_test_img, y_test_cat), verbose=1)
```

```python
acc_percp = perceptron.evaluate(X_test_img, y_test_cat,verbose=0)[1]
```

```python
acc_percp
```

```python
#ANN
ann = Sequential([
    Flatten(input_shape=(28,28)),
    Dense(128, activation="relu"),
    Dense(64, activation="relu"),
    Dense(10, activation="softmax")
])
```

```python
ann.compile(optimizer="adam" , loss="categorical_crossentropy", metrics=["accuracy"])
```

```python
history_ann=ann.fit(X_train_img,y_train_cat,epochs=5,batch_size=32,validation_data=(X_test_img,y_test_cat),verbose=1)
```

```python
acc_ann=ann.evaluate(X_test_img,y_test_cat,verbose=0)[1]
```

```python
X_train_cnn = X_train.reshape(-1, 28, 28, 1)
X_test_cnn = X_test.reshape(-1, 28, 28, 1)
```

```python
cnn = Sequential([
    Conv2D(32, kernel_size=(3,3), activation="relu", input_shape=(28,28,1)),
    MaxPooling2D(pool_size=(2,2)),
    Conv2D(64, kernel_size=(3,3), activation="relu"),
    MaxPooling2D(pool_size=(2,2)),
    Flatten(),
    Dense(128, activation="relu"),
    Dropout(0.5),
    Dense(10, activation="softmax")
])
```

```python
cnn.compile(optimizer="adam", loss="categorical_crossentropy", metrics=["accuracy"])
```

```python
history_cnn=cnn.fit(X_train_cnn,y_train_cat,epochs=5,batch_size=32,validation_data=(X_test_cnn,y_test_cat),verbose=1)
```
